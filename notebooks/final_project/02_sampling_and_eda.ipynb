## Sampling interpretation

- The job catalog uses a stable hash sample keyed by `job_link`; 1%, 5%, and 10% runs are scale experiments.
- Resume rows are not randomly sampled before splitting. Normalized job-description groups are assigned together to train, validation, or test.
- Development benchmarks use validation queries. The test set is not used to select features, thresholds, or models.
