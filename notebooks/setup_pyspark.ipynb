```python
import os, sys

os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable
# TODO: Replace this placeholder with the local Hadoop installation path.
HADOOP_PATH = r"C:\path\to\hadoop"
os.environ["HADOOP_HOME"] = HADOOP_PATH
os.environ["hadoop.home.dir"] = HADOOP_PATH
os.environ["PATH"] = os.path.join(HADOOP_PATH, "bin") + os.pathsep + os.environ["PATH"]

# TODO: Replace <YOUR_USERNAME> or set these paths for your AWS profile files.
AWS_CONFIG_DIR = r"C:\Users\<YOUR_USERNAME>\.aws"
os.environ["AWS_SHARED_CREDENTIALS_FILE"] = os.path.join(AWS_CONFIG_DIR, "credentials")
os.environ["AWS_CONFIG_FILE"] = os.path.join(AWS_CONFIG_DIR, "config")
os.environ["AWS_PROFILE"] = "default"

from pyspark.sql import SparkSession
import re

spark = (
    SparkSession.builder
    .appName("DS5110-S3")
    .config("spark.jars.packages",
            "org.apache.hadoop:hadoop-aws:3.3.4,com.amazonaws:aws-java-sdk-bundle:1.12.262")
    .config("spark.hadoop.fs.s3a.impl",
            "org.apache.hadoop.fs.s3a.S3AFileSystem")
    .config("spark.hadoop.fs.s3a.aws.credentials.provider",
            "com.amazonaws.auth.profile.ProfileCredentialsProvider")
    .config("spark.hadoop.fs.s3a.aws.credentials.provider.path",
            os.environ["AWS_SHARED_CREDENTIALS_FILE"])
    .config("spark.hadoop.fs.s3a.endpoint", "s3.us-east-2.amazonaws.com")
    .getOrCreate()
)
```
